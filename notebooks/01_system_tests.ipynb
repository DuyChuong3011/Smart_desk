### Test 1: Kiểm tra Camera và Vùng quan tâm (ROI)

```python
import cv2
import yaml

with open('../config/config.yaml', 'r') as f:
    config = yaml.safe_load(f)

cam_config = config['camera']
cap = cv2.VideoCapture(cam_config['index'])

if not cap.isOpened():
    print('Cannot open camera')
else:
    cap.set(cv2.CAP_PROP_FRAME_WIDTH, cam_config['width'])
    cap.set(cv2.CAP_PROP_FRAME_HEIGHT, cam_config['height'])
    cap.set(cv2.CAP_PROP_FPS, cam_config['fps'])
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        # Draw ROI for testing
        roi = config['desk']['roi']
        cv2.rectangle(frame, (roi['x1'], roi['y1']), (roi['x2'], roi['y2']), (0, 255, 0), 2)
        
        cv2.imshow('Camera Test (Press q to exit)', frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
```

### Test 2: Kiểm tra mô hình AI (YOLO)

```python
from ultralytics import YOLO
import cv2
import yaml

with open('../config/config.yaml', 'r') as f:
    config = yaml.safe_load(f)

model = YOLO(config['detection']['model_path'])
cam_config = config['camera']

cap = cv2.VideoCapture(cam_config['index'])
if not cap.isOpened():
    print('Cannot open camera')
else:
    cap.set(cv2.CAP_PROP_FRAME_WIDTH, cam_config['width'])
    cap.set(cv2.CAP_PROP_FRAME_HEIGHT, cam_config['height'])
    
    print("Press 'q' to exit")
    while True:
        ret, frame = cap.read()
        if not ret:
            break
            
        # Run YOLO on frame
        results = model(frame, verbose=False)
        annotated_frame = results[0].plot()
        
        cv2.imshow('YOLO Detection (Press q to exit)', annotated_frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
```
